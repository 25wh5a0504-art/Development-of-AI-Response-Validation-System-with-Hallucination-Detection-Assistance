## 1. Reverse a String

Slicing with a step of `-1` walks the string backwards.

```python
text = "interview"
print("Reversed string:", text[::-1])
```

## 2. Palindrome Check

A string is a palindrome if it reads the same forwards and backwards.

```python
text = "madam"

if text == text[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")
```

## 3. Factorial of a Number

Multiply every integer from 1 to `n`.

```python
n = 5

factorial = 1
for i in range(1, n + 1):
    factorial *= i

print("Factorial:", factorial)
```

## 4. Fibonacci Series

Each term is the sum of the previous two, starting from 0 and 1.

```python
n = 10

a, b = 0, 1

for _ in range(n):
    print(a, end=" ")
    a, b = b, a + b
```

## 5. Prime Number Check

Only test divisors up to the square root of `n`; if none divide it, it is prime.

```python
n = 29

if n < 2:
    print("Not prime")
else:
    is_prime = True

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            is_prime = False
            break

    print("Prime" if is_prime else "Not prime")
```

## 6. Largest and Second Largest

Convert to a set to drop duplicates, sort, then take the last two values.

```python
numbers = [10, 20, 4, 45, 99, 99, 45]

unique_numbers = list(set(numbers))

if len(unique_numbers) < 2:
    print("At least two different numbers are required.")
else:
    unique_numbers.sort()
    print("Largest:", unique_numbers[-1])
    print("Second largest:", unique_numbers[-2])
```

## 7. Remove Duplicates from a List

Keep only the first occurrence of each element, preserving order.

```python
numbers = [1, 2, 2, 3, 4, 4, 5]

result = []

for number in numbers:
    if number not in result:
        result.append(number)

print("Without duplicates:", result)
```

## 8. Frequency of Each Element

Use a dictionary and `get()` with a default of 0 to count occurrences.

```python
numbers = [1, 2, 2, 3, 3, 3, 4]

frequency = {}

for number in numbers:
    frequency[number] = frequency.get(number, 0) + 1

print(frequency)
```

## 9. Find Duplicate Elements

Track seen values in a set; any value seen twice is a duplicate.

```python
numbers = [1, 2, 3, 2, 4, 5, 1, 6]

seen = set()
duplicates = set()

for number in numbers:
    if number in seen:
        duplicates.add(number)
    else:
        seen.add(number)

print("Duplicate elements:", list(duplicates))
```

## 10. Bubble Sort

Repeatedly swap adjacent elements that are out of order; the largest value "bubbles" to the end each pass.

```python
numbers = [64, 34, 25, 12, 22, 11, 90]

n = len(numbers)

for i in range(n):
    for j in range(0, n - i - 1):
        if numbers[j] > numbers[j + 1]:
            numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]

print("Sorted list:", numbers)
```
