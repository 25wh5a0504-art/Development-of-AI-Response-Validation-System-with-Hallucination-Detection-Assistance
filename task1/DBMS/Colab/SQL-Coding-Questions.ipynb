# SQL Coding Questions

Important SQL interview queries, ready to run in Google Colab using SQLite (no setup needed).

**How to use:** run the setup cell first, then run any query cell. Or use **Runtime → Run all**.

## Setup: create the table and sample data

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect(":memory:")

conn.executescript("""
CREATE TABLE Employees (
    id INT PRIMARY KEY,
    name VARCHAR(100),
    department VARCHAR(100),
    salary DECIMAL(10, 2)
);

INSERT INTO Employees (id, name, department, salary) VALUES
(1, 'Asha', 'CSE', 50000),
(2, 'Ravi', 'ECE', 45000),
(3, 'Priya', 'CSE', 60000),
(4, 'Arun', 'EEE', 40000),
(5, 'Meena', 'CSE', 60000);
""")

def run(query):
    """Run a SQL query and show the result as a table."""
    return pd.read_sql_query(query, conn)

run("SELECT * FROM Employees")
```

## 1. Display employees from CSE

```python
run("""
SELECT * FROM Employees
WHERE department = 'CSE';
""")
```

## 2. Sort employees by salary (highest first)

```python
run("""
SELECT * FROM Employees
ORDER BY salary DESC;
""")
```

## 3. Find average salary

```python
run("""
SELECT AVG(salary) AS average_salary
FROM Employees;
""")
```

## 4. Count employees by department

```python
run("""
SELECT department, COUNT(*) AS employee_count
FROM Employees
GROUP BY department;
""")
```

## 5. Find highest salary

```python
run("""
SELECT MAX(salary) AS highest_salary
FROM Employees;
""")
```

## 6. Find second highest distinct salary

```python
run("""
SELECT MAX(salary) AS second_highest_salary
FROM Employees
WHERE salary < (SELECT MAX(salary) FROM Employees);
""")
```

## 7. Find duplicate salaries

```python
run("""
SELECT salary, COUNT(*) AS count
FROM Employees
GROUP BY salary
HAVING COUNT(*) > 1;
""")
```

## 8. Highest salary in each department

```python
run("""
SELECT department, MAX(salary) AS highest_salary
FROM Employees
GROUP BY department;
""")
```

## 9. Top N salaries (here N = 3)

```python
run("""
SELECT DISTINCT salary
FROM Employees
ORDER BY salary DESC
LIMIT 3;
""")
```

## 10. Employees whose name starts with a letter (here 'A')

```python
run("""
SELECT * FROM Employees
WHERE name LIKE 'A%';
""")
```
